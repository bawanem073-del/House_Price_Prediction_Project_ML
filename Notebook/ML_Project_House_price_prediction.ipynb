# 4. Data Cleaning

4.1 checking Outliers

```python
num_features = df.select_dtypes(include = "number")
```

```python
Q1 = num_features.quantile(0.25)
Q3 = num_features.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = (num_features < lower_bound) | (num_features > upper_bound)
print(outliers.sum())
```

4.2 Remove Outlier



```python
num_features = df.select_dtypes(include="number").columns

for col in num_features:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Clip values so anything outside the boundary gets mapped to the boundary line
    df[col] = df[col].clip(lower=lower_bound, upper=upper_bound)
```

4.3 Handlig Missing values

```python
df.isnull().sum()
```

```python
df['Car'] = df.groupby('Type')['Car'].transform(lambda x: x.fillna(x.median()))

df['CouncilArea'] = df.groupby('Suburb')['CouncilArea'].transform(
    lambda x: x.fillna(x.mode()[0] if not x.mode().empty else 'Unknown')
)

df['BuildingArea'] = df.groupby(['Type', 'Rooms'])['BuildingArea'].transform(
    lambda x: x.fillna(x.median())
)


df['YearBuilt'] = df.groupby('Suburb')['YearBuilt'].transform(
    lambda x: x.fillna(x.median())
)

df['BuildingArea'] = df['BuildingArea'].fillna(df['BuildingArea'].median())
df['YearBuilt'] = df['YearBuilt'].fillna(df['YearBuilt'].median())
```

# 8. Train Test Split

```python
df_new.select_dtypes(include = 'number').columns
```

```python
df_new.select_dtypes(include = 'object').columns
```

```python
numeric_columns = ['Rooms', 'Bathroom', 'Car', 'Distance', 'BuildingArea', 'Landsize','Age']
normal_columns = ['Grouped_Suburb', 'Grouped_SellerG']
ordinal_columns = ['Type']
```

```python
numeric_columns
```

```python
features_col = numeric_columns + normal_columns + ordinal_columns
X = df[features_col]
y = df['Price']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y ,test_size = 0.3, random_state = 42)
```

# 9. Preprocessing & ColumnTransformation

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder, StandardScaler
```

**Pipelins**

9.1 Numeric Pipline

```python
numeric_pipeline = Pipeline(steps = [('scaler', StandardScaler())])
```

9.2 Ordinal Pipeline

```python
ordinal_pipeline = Pipeline(steps = [('encoder_1',OrdinalEncoder(categories=[['u', 't', 'h']]))])
```

9.3 Normal pipeline

```python
normal_pipeline = Pipeline(steps = [('encoder_2', OneHotEncoder(handle_unknown='ignore'))])
```

**ColumnTransformer**

```python
from sklearn.compose import ColumnTransformer
```

```python
preprocessor = ColumnTransformer(transformers=[('numeric_pipeline', numeric_pipeline, numeric_columns),
                                               ('ordinal_pipeline', ordinal_pipeline, ordinal_columns),
                                               ('normal_pipeline', normal_pipeline, normal_columns)])
```

```python
preprocessor
```

# 10. Model Building

**10.1. Linear Regression**

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error, root_mean_squared_error

lr_pipeline = Pipeline(steps = [('preprocessor', preprocessor),
                                ('regressor', LinearRegression())])

lr_pipeline.fit(X_train, y_train)
# test prediction
lr_test_pred = lr_pipeline.predict(X_test)
# train prediction
lr_train_pred = lr_pipeline.predict(X_train)

lr_r2_score_test = r2_score(y_test, lr_test_pred)
lr_r2_score_train = r2_score(y_train, lr_train_pred)

lr_mae_test = mean_absolute_error(y_test, lr_test_pred)
lr_mae_train = mean_absolute_error(y_train, lr_train_pred)

lr_rmse_test = root_mean_squared_error(y_test, lr_test_pred)
lr_rmse_train = root_mean_squared_error(y_train, lr_train_pred)

print(f'R2 Score (Test): {lr_r2_score_test}')
print(f'R2 Score (Train): {lr_r2_score_train}')
print(f'MAE (Test): {lr_mae_test}')
print(f'MAE (Train): {lr_mae_train}')
```

**10.2. Random Forest regressor**

```python
from sklearn.ensemble import RandomForestRegressor

rf_pipeline = Pipeline(steps = [('preprocessor', preprocessor),
                                ('random forest', RandomForestRegressor())])
rf_pipeline.fit(X_train, y_train)

rf_test_pred = rf_pipeline.predict(X_test)
rf_train_pred = rf_pipeline.predict(X_train)

rf_r2_score_test = r2_score(y_test, rf_test_pred)
rf_r2_score_train = r2_score(y_train, rf_train_pred)

rf_mae_test = mean_absolute_error(y_test, rf_test_pred)
rf_mae_train = mean_absolute_error(y_train, rf_train_pred)

rf_rmse_test = root_mean_squared_error(y_test, rf_test_pred)
rf_rmse_train = root_mean_squared_error(y_train, rf_train_pred)

print(f"R2 Score (Test): {rf_r2_score_test}")
print(f"R2 Score (Train): {rf_r2_score_train}")
```

**10.3. Gredint Boosting Regressor**

```python
from sklearn.ensemble import GradientBoostingRegressor

gbr_pipeline = Pipeline(steps = [('preprocessor', preprocessor),
                                 ('Gredient Boost', GradientBoostingRegressor())])

gbr_pipeline.fit(X_train, y_train)

gbr_test_pred = gbr_pipeline.predict(X_test)
gbr_train_pred = gbr_pipeline.predict(X_train)

gbr_r2_score_test = r2_score(y_test, gbr_test_pred)
gbr_r2_score_train = r2_score(y_train, gbr_train_pred)

gbr_mae_test = mean_absolute_error(y_test, gbr_test_pred)
gbr_mae_train = mean_absolute_error(y_train, gbr_train_pred)

gbr_rmse_test = root_mean_squared_error(y_test, gbr_test_pred)
gbr_rmse_train = root_mean_squared_error(y_train, gbr_train_pred)

print(f"R2 Score (Test): {gbr_r2_score_test}")
print(f"R2 Score (Train): {gbr_r2_score_train}")
```

**10.4 Xgboost Regressor**

```python
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error, root_mean_squared_error
xgb_pipeline = Pipeline(steps = [('preprocessor', preprocessor),
                                 ('Xgboost Regressor', XGBRegressor())])

xgb_pipeline.fit(X_train, y_train)
xgb_test_pred = xgb_pipeline.predict(X_test)
xgb_train_pred = xgb_pipeline.predict(X_train)

xgb_r2_score_test = r2_score(y_test, xgb_test_pred)
xgb_r2_score_train = r2_score(y_train, xgb_train_pred)

xgb_mae_test = mean_absolute_error(y_test, xgb_test_pred)
xgb_mae_train = mean_absolute_error(y_train, xgb_train_pred)

xgb_rmse_test = root_mean_squared_error(y_test, xgb_test_pred)
xgb_rmse_train = root_mean_squared_error(y_train, xgb_train_pred)

print(f"R2 Score (Test): {xgb_r2_score_test}")
print(f"R2 Score (Train): {xgb_r2_score_train}")

```

**10.5 Hyperparameter Tunnig the Xgboost**

```python
from sklearn.model_selection import RandomizedSearchCV

param_grid = {
    "Xgboost Regressor__n_estimators": [100, 200, 300, 500],
    "Xgboost Regressor__learning_rate": [0.01, 0.05, 0.1, 0.2],
    "Xgboost Regressor__max_depth": [3, 4, 5, 6, 8],
    "Xgboost Regressor__subsample": [0.6, 0.7, 0.8, 0.9, 1.0],
    "Xgboost Regressor__colsample_bytree": [0.6, 0.7, 0.8, 0.9, 1.0],
    "Xgboost Regressor__gamma": [0, 0.1, 0.2, 0.5],
    "Xgboost Regressor__reg_alpha": [0, 0.01, 0.1, 1],
    "Xgboost Regressor__reg_lambda": [0.1, 1, 5, 10]
}


xgb_random_search = RandomizedSearchCV(
    estimator = xgb_pipeline,
    param_distributions= param_grid,
    n_iter = 30,
    scoring= 'r2',
    cv = 5,
    verbose = 1,
    random_state=42,
    n_jobs = -1
)


xgb_random_search.fit(X_train, y_train)

# Output best hyperparameter combination & score
print("Best Parameters:", xgb_random_search.best_params_)
print("Best CV Score (R2):", xgb_random_search.best_score_)
```

```python
xgb_best_pipeline = xgb_random_search.best_estimator_

xgb_best_test_pred = xgb_best_pipeline.predict(X_test)
xgb_best_train_pred = xgb_best_pipeline.predict(X_train)

xgb_best_r2_score_test = r2_score(y_test, xgb_best_test_pred)
xgb_best_r2_score_train = r2_score(y_train, xgb_best_train_pred)

xgb_best_mae_test = mean_absolute_error(y_test, xgb_best_test_pred)
xgb_best_mae_train = mean_absolute_error(y_train, xgb_best_train_pred)

xgb_best_rmse_test = root_mean_squared_error(y_test, xgb_best_test_pred)
xgb_best_rmse_train = root_mean_squared_error(y_train, xgb_best_train_pred)

```
